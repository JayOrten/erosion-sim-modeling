# C5_band2. Scale: band 2

Band 2 (period 64 to 128): two levels with a 128-cell coarse window, or three levels?

See `docs/plans/experiments-plan.md`, entry C5_band2. Results: `artifacts/results/world_band2/`.

## Run

From the repo root. `uv run erosion-sim list` shows what is already trained and evaluated.

**1. Data.** Skip any that already exist.

Local:

```sh
(cd datagen && uv run erosion-datagen generate erosion_master/main_1024)
```

Cluster:

```sh
sbatch slurm/datagen.sbatch erosion_master/main_1024
```

**2. Train**, in this order. Each resumes if stopped; a finished model is skipped.

Local:

```sh
uv run erosion-sim train band2_two_4m
uv run erosion-sim train band2_two_1m
uv run erosion-sim train band2_three_16m
uv run erosion-sim train band2_three_4m
uv run erosion-sim train band2_three_1m
```

Cluster:

```sh
sbatch slurm/train.sbatch band2_two_4m
sbatch slurm/train.sbatch band2_two_1m
sbatch slurm/train.sbatch band2_three_16m
sbatch slurm/train.sbatch band2_three_4m
sbatch slurm/train.sbatch band2_three_1m
```

**3. Evaluate**, once training has finished. It refuses to replace existing results.

Local:

```sh
uv run erosion-sim evaluate world_band2 band2_two
uv run erosion-sim evaluate world_band2 band2_three
uv run erosion-sim evaluate world_band2 no_change
uv run erosion-sim evaluate world_band2 analytical
uv run erosion-sim evaluate world_band2 analytical_chunks_256
uv run erosion-sim evaluate world_band2 analytical_blended_256
```

Cluster:

```sh
sbatch slurm/evaluate.sbatch world_band2 band2_two
sbatch slurm/evaluate.sbatch world_band2 band2_three
sbatch slurm/evaluate.sbatch world_band2 no_change
sbatch slurm/evaluate.sbatch world_band2 analytical
sbatch slurm/evaluate.sbatch world_band2 analytical_chunks_256
sbatch slurm/evaluate.sbatch world_band2 analytical_blended_256
```

Then run the cells below.

How to read the tables and plots: [`docs/guides/world.md`](../../docs/guides/world.md).

````python
import json

from erosion_sim.erosion import plots
from erosion_sim.evaluation.results import load
from erosion_sim.paths import model_dir
from erosion_sim.training.train import plot_history

EVALUATION = "world_band2"  # configs/evaluations/world_band2.yaml
CASCADES = ["band2_two", "band2_three"]  # configs/predictors/
BASELINES = ["no_change", "analytical", "analytical_chunks_256", "analytical_blended_256"]  # configs/predictors/
PREDICTORS = CASCADES + BASELINES
MODELS = ["band2_two_4m", "band2_two_1m", "band2_three_16m", "band2_three_4m", "band2_three_1m"]  # every trained model these use, for the training plots

results = load(EVALUATION, *PREDICTORS)
{p: {k: m[k] for k in ("time", "code", "slurm_job")} for p, m in results.manifest["predictors"].items()}
````

## Training

````python
for name in MODELS:
    print(name, model_dir(name))
    plot_history(json.loads((model_dir(name) / "history.json").read_text()))
````

## Cascades, levels alone, and baselines, on the runs' interiors

````python
plots.table(results, plots.TABLE + ('chunk_edge_rmse', 'seconds_per_query'))
````

````python
if "dT" in results.scalars["by"]:
    plots.by_table(results, "dT", "skill", kinds=("cascade", "baseline"))
````

## Seams

How much neighboring windows disagree before blending, per level and phase, as a share of the change.

````python
for world, levels in results.scalars["seams"].items():
    for level, phases in levels.items():
        print(world, level, [f"{p['relative']:.1%}" for p in phases])
````
